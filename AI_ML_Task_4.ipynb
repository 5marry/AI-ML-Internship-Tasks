# **Objective:**
The primary goal of this task is to develop a conversational AI agent capable of providing clear and friendly responses to general health-related inquiries. This task focuses on the practical application of Prompt Engineering—the art of crafting specific instructions to guide a Large Language Model (LLM) toward a desired behavior.

# **2. The Chatbot Code**

```python
from transformers import pipeline

# We stay with GPT-2 because it's fast, but we improve how it reads the prompt
chatbot = pipeline("text-generation", model="gpt2")

def health_chatbot(user_query):
    # Improved Prompt Engineering: We give a clearer example of how to answer
    prompt = (
        "Instructions: Act as a helpful medical assistant. Provide a brief, logical answer. "
        "Do not talk about unrelated topics.\n\n"
        f"Question: {user_query}\n"
        "Answer:"
    )

    # We use 'temperature' to make it less random and 'pad_token_id' to fix errors
    response = chatbot(
        prompt,
        max_new_tokens=50,
        temperature=0.7,
        num_return_sequences=1,
        pad_token_id=50256,
        truncation=True
    )

    # Extract only the answer part
    full_text = response[0]['generated_text']
    answer = full_text.split("Answer:")[-1].strip()

    disclaimer = "\n\n(Disclaimer: I am an AI, not a doctor. Please consult a medical professional.)"
    return answer + disclaimer

# Test it again
print(health_chatbot("What causes a sore throat?"))
```

# **EXPLANATION:**
This task uses Prompt Engineering to transform a general-purpose Large Language Model (LLM) into a specialized medical assistant.

 The code works in three main steps:

 **1.Model Loading:** We initialize a text-generation pipeline using a pre-trained model (like GPT-2) from the Hugging Face library.

**2.Prompt Design:** A "System Prompt" is created to give the AI a specific persona. By instructing it to "Act like a helpful medical assistant," we guide the model to provide clear, relevant answers rather than random text.

**3.Safety & Controls:**

**Disclaimer:** A mandatory safety filter is added to every response to ensure the user knows the AI is not a substitute for a real doctor.

**Parameters:** We set max_new_tokens and temperature to keep the answers concise and prevent the model from "hallucinating" or rambling
