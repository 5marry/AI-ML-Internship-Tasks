### **Task 1: Exploring and Visualizing the Iris Dataset**

**Objective:** The goal of this task is to load the Iris dataset and use exploratory data analysis (EDA) to understand the relationships between different flower features (sepal and petal measurements) and identify any patterns or outliers.

**1.data Load**

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# Load the dataset using seaborn (standard way to get Iris) [cite: 11, 13]
df = sns.load_dataset('iris')

# Print the shape and column names [cite: 14]
print("Dataset Shape:", df.shape)
print("Column Names:", df.columns.tolist())

# Show the first few rows [cite: 14]
df.head()
```

**2. Descriptive Statistics**

```python
# Detailed information about data types and missing values [cite: 15]
print(df.info())

# Summary statistics for numerical columns [cite: 15, 23]
df.describe()
```

**3. Data Visualization**

```python
# 1. Scatter Plot: Showing relationship between features (Sepal Length vs Width) [cite: 17]
plt.figure(figsize=(8, 5))
sns.scatterplot(data=df, x='sepal_length', y='sepal_width', hue='species')
plt.title('Sepal Length vs Sepal Width')
plt.show()

# 2. Histograms: Showing value distributions for each feature [cite: 18]
df.hist(figsize=(10, 8), bins=20, color='skyblue', edgecolor='black')
plt.suptitle('Distribution of Iris Features')
plt.show()

# 3. Box Plots: Identifying outliers in the data [cite: 19]
plt.figure(figsize=(10, 6))
sns.boxplot(data=df)
plt.title('Box Plot of Iris Features (Outlier Detection)')
plt.show()
```

# **1.Scatter Plot**

**Scatter Plot Analysis**
The scatter plot shows the relationship between sepal length and sepal width, colored by species.

**Insight:** The Setosa species (blue dots) forms a very distinct cluster with shorter sepal lengths but wider sepal widths compared to the other two species.

Versicolor and Virginica overlap significantly in their sepal measurements, suggesting that sepal dimensions alone might not be enough to distinguish them perfectly.

# **2.Histograms**

**Distribution Analysis (Histograms)**
* The histograms show how the values for each feature are spread out.

**Insight:** Petal length and petal width show a "bimodal" distribution (two distinct peaks). This confirms that there is a clear gap between one group of flowers (likely Setosa) and the others.

Sepal length and width follow a more "normal" or bell-shaped distribution.

# **3.Box Plot**

**Outlier Detection (Box Plot)**
* Box plots are used to identify the median, range, and potential outliers in the data.

**Insight:** Looking at the "sepal_width" column, we can see a few individual circles above and below the whiskers. These represent outliers—data points that are unusually high or low compared to the rest of the group.

The other features (sepal_length, petal_length, and petal_width) appear to have no significant outliers.

# **SUMMARY:**
After exploring the dataset, it is clear that petal measurements are much better "predictors" for identifying the species than sepal measurements. The data is clean, with only minor outliers in the sepal width category. This foundational understanding is essential before building any machine learning models
